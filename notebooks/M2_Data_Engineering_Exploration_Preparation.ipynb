```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    return {
        "variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "borne_inférieure": lower_bound,
        "borne_supérieure": upper_bound,
        "nombre_outliers": len(outliers)
    }


outlier_results = []

for column in ["Age", "Total_Spent"]:
    outlier_results.append(
        detect_outliers_iqr(customers, column)
    )

outlier_results.append(
    detect_outliers_iqr(products, "Price")
)

for column in ["Quantity", "Sale_Price"]:
    outlier_results.append(
        detect_outliers_iqr(sales, column)
    )

for column in ["Budget", "Impressions", "Clicks", "Conversions"]:
    outlier_results.append(
        detect_outliers_iqr(marketing, column)
    )

outliers_df = pd.DataFrame(outlier_results)

display(outliers_df)
```

```python
print("===== INSPECTION DES OUTLIERS =====")

# -------------------------
# Age
# -------------------------
print("\n--- OUTLIERS AGE ---")

Q1 = customers["Age"].quantile(0.25)
Q3 = customers["Age"].quantile(0.75)
IQR = Q3 - Q1

age_outliers = customers[
    (customers["Age"] < Q1 - 1.5 * IQR) |
    (customers["Age"] > Q3 + 1.5 * IQR)
]

display(
    age_outliers[
        ["Customer_ID", "Name", "Age", "Gender",
         "Location", "Join_Date", "Total_Spent"]
    ].sort_values("Age")
)


# -------------------------
# Total_Spent
# -------------------------
print("\n--- OUTLIERS TOTAL_SPENT ---")

Q1 = customers["Total_Spent"].quantile(0.25)
Q3 = customers["Total_Spent"].quantile(0.75)
IQR = Q3 - Q1

spent_outliers = customers[
    (customers["Total_Spent"] < Q1 - 1.5 * IQR) |
    (customers["Total_Spent"] > Q3 + 1.5 * IQR)
]

print("Nombre :", len(spent_outliers))

display(
    spent_outliers[
        ["Customer_ID", "Name", "Age",
         "Gender", "Location", "Total_Spent"]
    ].sort_values("Total_Spent", ascending=False).head(20)
)


# -------------------------
# Clicks
# -------------------------
print("\n--- OUTLIERS CLICKS ---")

Q1 = marketing["Clicks"].quantile(0.25)
Q3 = marketing["Clicks"].quantile(0.75)
IQR = Q3 - Q1

click_outliers = marketing[
    (marketing["Clicks"] < Q1 - 1.5 * IQR) |
    (marketing["Clicks"] > Q3 + 1.5 * IQR)
]

display(click_outliers)


# -------------------------
# Conversions
# -------------------------
print("\n--- OUTLIERS CONVERSIONS ---")

Q1 = marketing["Conversions"].quantile(0.25)
Q3 = marketing["Conversions"].quantile(0.75)
IQR = Q3 - Q1

conversion_outliers = marketing[
    (marketing["Conversions"] < Q1 - 1.5 * IQR) |
    (marketing["Conversions"] > Q3 + 1.5 * IQR)
]

display(conversion_outliers)
```

```python
# ============================================================
# 5. CONTROLES METIER MARKETING
# ============================================================

print("===== CONTROLES METIER MARKETING =====")

print(
    "Budget <= 0 :",
    (marketing_clean["budget"] <= 0).sum()
)

print(
    "Impressions < 0 :",
    (marketing_clean["impressions"] < 0).sum()
)

print(
    "Clicks < 0 :",
    (marketing_clean["clicks"] < 0).sum()
)

print(
    "Conversions < 0 :",
    (marketing_clean["conversions"] < 0).sum()
)

print(
    "Clicks > Impressions :",
    (marketing_clean["clicks"] > marketing_clean["impressions"]).sum()
)

print(
    "Conversions > Clicks :",
    (marketing_clean["conversions"] > marketing_clean["clicks"]).sum()
)

print(
    "Start_Date > End_Date :",
    (marketing_clean["start_date"] > marketing_clean["end_date"]).sum()
)

print(
    "Dates invalides :",
    (
        marketing_clean["start_date"].isna() |
        marketing_clean["end_date"].isna()
    ).sum()
)
```
